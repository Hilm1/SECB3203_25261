```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, LabelEncoder

# 1. Importing Dataset
df = pd.read_csv('data/breast_cancer_data.csv')

# 2. Understanding the data (initial check)
print(df.info())

# 3. Identifying and handling missing values
# Dropping 'Unnamed: 32' as it is usually an empty trailing column in Kaggle
df.drop(['id', 'Unnamed: 32'], axis=1, inplace=True, errors='ignore')
print("Missing values after cleaning:", df.isnull().sum().sum())

# 4. Indicator variables (Label Encoding)
# M = 1, B = 0
le = LabelEncoder()
df['diagnosis'] = le.fit_transform(df['diagnosis'])

# 5. Data Normalization (Scaling)
# Separate features (X) and target (y)
X = df.drop('diagnosis', axis=1)
y = df['diagnosis']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert back to DataFrame for easier inspection
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)
print("Data scaling completed successfully.")
```
